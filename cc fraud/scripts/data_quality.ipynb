```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

### Checking normal distibution of features

```python
df = pd.read_csv('../dataset/creditcard.csv')
df = df.drop(columns=['Time', 'Class'])

scaler = StandardScaler()

data = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)

column_num = len(data.columns)

for c in range(column_num):
    try:
        plt.hist(data.iloc[:, c], bins=500)
        plt.title(f'Histogram of {data.columns[c]}')
        plt.show()
    except Exception as e:
        print(f'Error occurred while plotting {data.columns[c]}: {e}')
```
